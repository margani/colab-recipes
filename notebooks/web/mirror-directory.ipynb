```python
#@title Download
url = "" #@param {type:"string"}
folder = "Mirror" #@param {type:"string"}

import os, subprocess, urllib.parse
if not url:
    raise SystemExit("Enter a directory URL.")
url = url.rstrip("/") + "/"
depth = len([p for p in urllib.parse.urlparse(url).path.split("/") if p])
dest_dir = f"/content/drive/MyDrive/{folder}"
os.makedirs(dest_dir, exist_ok=True)
# Argument lists, not ! lines: the shell would run $(...) or backticks found in a URL.
subprocess.run(["wget", "-r", "-np", "-nH", f"--cut-dirs={depth}", "-c", "-e", "robots=off", "-R", "index.html*",
                "--reject-regex", "[?]C=", "--no-verbose", "-P", dest_dir, "--", url])
subprocess.run(["ls", "-lh", "--", dest_dir]);
```
