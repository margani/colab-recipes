# Verify SHA256

```python
#@title Connect Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#@title Verify
file_path = "/content/drive/MyDrive/Downloads/" #@param {type:"string"}
expected_sha256 = "" #@param {type:"string"}

import hashlib
h = hashlib.sha256()
with open(file_path, "rb") as f:
    for chunk in iter(lambda: f.read(8 * 1024 * 1024), b""):
        h.update(chunk)
actual = h.hexdigest()
print("SHA256:", actual)
if expected_sha256:
    print("✅ Match" if actual.lower() == expected_sha256.strip().lower() else "❌ Mismatch: the file is corrupt or not the expected one")
```
