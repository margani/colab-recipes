```python
#@title Extract
file_path = "" #@param {type:"string"}
audio_format = "mp3" #@param ["mp3", "m4a", "wav"]

import os, subprocess
if not file_path:
    raise SystemExit("Enter a file path.")
if not os.path.isfile(file_path):
    raise SystemExit(f"Not a file: {file_path}")
base = os.path.splitext(file_path)[0]
output = f"{base}.{audio_format}"
if output == file_path:
    output = f"{base}-audio.{audio_format}"
codec = {"mp3": ["-c:a", "libmp3lame", "-q:a", "2"],
         "m4a": ["-c:a", "aac", "-b:a", "192k"],
         "wav": ["-c:a", "pcm_s16le"]}[audio_format]
subprocess.run(["ffmpeg", "-hide_banner", "-loglevel", "error", "-stats", "-y",
                "-i", file_path, "-vn", *codec, output], check=True)
!ls -lh "{output}"
```
