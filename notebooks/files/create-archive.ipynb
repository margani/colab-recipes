# Create archive

```python
#@title Connect Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#@title Create
source = "" #@param {type:"string"}
archive_format = "zip" #@param ["zip", "7z"]
password = "" #@param {type:"string"}
split_size = "" #@param {type:"string"}

# Accept paths as Drive shows them (Temp\video.mp4) or as full /content paths.
source = source.strip().strip('"').replace("\\", "/")
if source and not source.startswith("/"):
    source = "/content/drive/MyDrive/" + source.removeprefix("MyDrive/").removeprefix("My Drive/")

import os, glob, subprocess
if not source:
    raise SystemExit("Enter a file or folder path.")
source = source.rstrip("/")
if not os.path.exists(source):
    raise SystemExit(f"Not found: {source}")
output = f"{source}.{archive_format}"
# Re-running must replace the old archive; 7z would otherwise add to it.
for old in [output] + glob.glob(glob.escape(output) + ".[0-9][0-9][0-9]"):
    if os.path.exists(old):
        os.remove(old)
args = [f"-t{archive_format}"]
if password:
    args += [f"-p{password}", "-mhe=on" if archive_format == "7z" else "-mem=AES256"]
if split_size:
    args.append(f"-v{split_size.lower()}")
!apt-get -qq install -y p7zip-full > /dev/null
subprocess.run(["7z", "a", *args, output, source], check=True, stdout=subprocess.DEVNULL)
print("✅ Done")
!ls -lh "{output}"*
```
