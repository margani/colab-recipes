```python
#@title Merge
folder = "" #@param {type:"string"}
output_name = "merged.pdf" #@param {type:"string"}

# Accept paths as Drive shows them (Temp\video.mp4) or as full /content paths.
folder = folder.strip().strip('"').replace("\\", "/")
if folder and not folder.startswith("/"):
    folder = "/content/drive/MyDrive/" + folder.removeprefix("MyDrive/").removeprefix("My Drive/")

import os, re
if not folder:
    raise SystemExit("Enter a folder path.")
if not os.path.isdir(folder):
    raise SystemExit(f"Not a folder: {folder}")
!pip install -q pypdf
from pypdf import PdfWriter
output = os.path.join(folder, output_name)
natural = lambda name: [int(t) if t.isdigit() else t.lower() for t in re.split(r"(\d+)", name)]
files = sorted((f for f in os.listdir(folder)
                if f.lower().endswith(".pdf") and os.path.join(folder, f) != output), key=natural)
if not files:
    raise SystemExit("No PDF files in this folder.")
writer = PdfWriter()
for f in files:
    print("Adding", f)
    writer.append(os.path.join(folder, f))
writer.write(output)
print("✅", output)
```
