```python
#@title Download
url = "" #@param {type:"string"}
folder = "Downloads" #@param {type:"string"}

import os, subprocess
if not url:
    raise SystemExit("Enter a Google Drive link.")
dest_dir = f"/content/drive/MyDrive/{folder}"
os.makedirs(dest_dir, exist_ok=True)
!pip install -q -U gdown
# Argument lists, not ! lines: the shell would run $(...) or backticks found in a link.
subprocess.run(["gdown", "-c", "-O", dest_dir + "/", "--", url])
subprocess.run(["ls", "-lh", "--", dest_dir]);
```
