```python
#@title Join
first_part = "" #@param {type:"string"}

import os, re, glob, shutil
m = re.fullmatch(r"(.+?)(\.part|\.)(\d+)", first_part)
if not m:
    raise SystemExit("Enter the first part, e.g. movie.mkv.part000 or movie.mkv.001.")
base, sep, num = m.groups()
parts = sorted(glob.glob(glob.escape(base + sep) + "[0-9]" * len(num)))
if not parts:
    raise SystemExit(f"No parts found for {base}")
with open(base, "wb") as out:
    for p in parts:
        print("Adding", os.path.basename(p))
        with open(p, "rb") as f:
            shutil.copyfileobj(f, out, 16 * 1024 * 1024)
print(f"✅ {base} ({len(parts)} parts)")
```
