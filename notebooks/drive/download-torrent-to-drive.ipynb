# Download torrent to Google Drive

```python
#@title Connect Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#@title Check free space
!df -h /content/drive | tail -1
```

```python
#@title Download
torrent = "" #@param {type:"string"}
folder = "Torrents" #@param {type:"string"}

# Accept paths as Drive shows them (Temp\video.mp4) or as full /content paths.
torrent = torrent.strip().strip('"').replace("\\", "/")
if torrent and not torrent.startswith(("/", "magnet:", "http")):
    torrent = "/content/drive/MyDrive/" + torrent.removeprefix("MyDrive/").removeprefix("My Drive/")

import os, subprocess
if not torrent:
    raise SystemExit("Enter a magnet link or a .torrent file path.")
dest_dir = f"/content/drive/MyDrive/{folder}"
os.makedirs(dest_dir, exist_ok=True)
!apt-get -qq install -y aria2 > /dev/null
subprocess.run(["aria2c", "-c", "--seed-time=0", "--file-allocation=none",
                "--summary-interval=30", "--console-log-level=warn",
                "-d", dest_dir, torrent], check=True)
!ls -lh "{dest_dir}"
```
