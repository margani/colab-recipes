## 3. Download a URL to Drive
The URL must be a direct link: opening it in a browser starts the download.

Leave the file name empty to take it from the URL.

```python
#@title Download file
url = "" #@param {type:"string"}
folder = "Downloads" #@param {type:"string"}
filename = "" #@param {type:"string"}

import os, urllib.parse
if not url:
    raise SystemExit("Enter a URL.")
if not filename:
    filename = os.path.basename(urllib.parse.urlparse(url).path) or "download.bin"
    filename = urllib.parse.unquote(filename)[:200]
dest_dir = f"/content/drive/MyDrive/{folder}"
os.makedirs(dest_dir, exist_ok=True)
dest = f"{dest_dir}/{filename}"
print("Saving to:", dest)
!wget -c -q --show-progress -O "{dest}" "{url}"
!ls -lh "{dest}"
```

## 4. Verify SHA256 (optional)
Compare against the checksum published by the source.

```python
#@title Verify SHA256
file_path = "/content/drive/MyDrive/Downloads/" #@param {type:"string"}
expected_sha256 = "" #@param {type:"string"}

import hashlib
h = hashlib.sha256()
with open(file_path, "rb") as f:
    for chunk in iter(lambda: f.read(8 * 1024 * 1024), b""):
        h.update(chunk)
actual = h.hexdigest()
print("SHA256:", actual)
if expected_sha256:
    print("✅ Match" if actual.lower() == expected_sha256.strip().lower() else "❌ Mismatch: the file is corrupt or not the expected one")
```

## 6. Rename a file (optional)

```python
#@title Rename file
old_path = "/content/drive/MyDrive/Downloads/" #@param {type:"string"}
new_name = "" #@param {type:"string"}

import os
new_path = os.path.join(os.path.dirname(old_path), new_name)
# Re-running after a successful rename must not crash on the missing old file.
if not os.path.exists(old_path) and os.path.exists(new_path):
    print("✅ Already renamed:", new_path)
else:
    os.rename(old_path, new_path)
    print("✅", new_path)
```
