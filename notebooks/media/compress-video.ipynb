# Compress video

```python
#@title Connect Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#@title Compress
file_path = "" #@param {type:"string"}
max_height = "720" #@param ["original", "1080", "720", "480"]
crf = 28 #@param {type:"slider", min:18, max:35, step:1}

# Accept paths as Drive shows them (Temp\video.mp4) or as full /content paths.
file_path = file_path.strip().strip('"').replace("\\", "/")
if file_path and not file_path.startswith("/"):
    file_path = "/content/drive/MyDrive/" + file_path.removeprefix("MyDrive/").removeprefix("My Drive/")

import os, subprocess
if not file_path:
    raise SystemExit("Enter a file path.")
if not os.path.isfile(file_path):
    raise SystemExit(f"Not a file: {file_path}")
output = os.path.splitext(file_path)[0] + "-small.mp4"
scale = [] if max_height == "original" else ["-vf", f"scale=-2:'min({max_height},ih)'"]
subprocess.run(["ffmpeg", "-hide_banner", "-loglevel", "error", "-stats", "-y", "-i", file_path, *scale,
                "-c:v", "libx264", "-preset", "medium", "-crf", str(crf),
                "-c:a", "aac", "-b:a", "128k", "-movflags", "+faststart", output], check=True)
!ls -lh "{file_path}" "{output}"
```
