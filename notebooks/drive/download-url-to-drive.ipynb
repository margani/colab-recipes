```python
#@title Download
url = "" #@param {type:"string"}
folder = "Downloads" #@param {type:"string"}
filename = "" #@param {type:"string"}

import os, subprocess, urllib.parse
if not url:
    raise SystemExit("Enter a URL.")
if not filename:
    # Decode before taking the last part, so an encoded "/" cannot leave the folder.
    filename = os.path.basename(urllib.parse.unquote(urllib.parse.urlparse(url).path))[:200]
    if filename in ("", ".", ".."):
        filename = "download.bin"
dest_dir = f"/content/drive/MyDrive/{folder}"
os.makedirs(dest_dir, exist_ok=True)
dest = f"{dest_dir}/{filename}"
print("Saving to:", dest)
# Argument lists, not ! lines: the shell would run $(...) or backticks found in a URL or file name.
subprocess.run(["wget", "-c", "-q", "--show-progress", "-O", dest, "--", url])
subprocess.run(["ls", "-lh", "--", dest]);
```
