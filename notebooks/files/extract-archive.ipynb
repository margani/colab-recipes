# Extract archive

```python
#@title Connect Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#@title Extract
archive_path = "/content/drive/MyDrive/Downloads/" #@param {type:"string"}
extract_to = "/content/drive/MyDrive/Downloads/extracted" #@param {type:"string"}

# Accept paths as Drive shows them (Temp\video.mp4) or as full /content paths.
archive_path = archive_path.strip().strip('"').replace("\\", "/")
if archive_path and not archive_path.startswith("/"):
    archive_path = "/content/drive/MyDrive/" + archive_path.removeprefix("MyDrive/").removeprefix("My Drive/")
extract_to = extract_to.strip().strip('"').replace("\\", "/")
if extract_to and not extract_to.startswith("/"):
    extract_to = "/content/drive/MyDrive/" + extract_to.removeprefix("MyDrive/").removeprefix("My Drive/")

!apt-get -qq install -y p7zip-full unrar > /dev/null
!mkdir -p "{extract_to}"
!7z x -y "{archive_path}" -o"{extract_to}" > /dev/null && echo "✅ Extracted"
!ls -lh "{extract_to}"
```
