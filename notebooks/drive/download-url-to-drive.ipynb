```python
#@title Download
url = "" #@param {type:"string"}
folder = "Downloads" #@param {type:"string"}
filename = "" #@param {type:"string"}

import os, urllib.parse
if not url:
    raise SystemExit("Enter a URL.")
if not filename:
    filename = os.path.basename(urllib.parse.urlparse(url).path) or "download.bin"
    filename = urllib.parse.unquote(filename)[:200]
dest_dir = f"/content/drive/MyDrive/{folder}"
os.makedirs(dest_dir, exist_ok=True)
dest = f"{dest_dir}/{filename}"
print("Saving to:", dest)
!wget -c -q --show-progress -O "{dest}" "{url}"
!ls -lh "{dest}"
```

```python
#@title Rename
old_path = "/content/drive/MyDrive/Downloads/" #@param {type:"string"}
new_name = "" #@param {type:"string"}

import os
new_path = os.path.join(os.path.dirname(old_path), new_name)
# Re-running after a successful rename must not crash on the missing old file.
if not os.path.exists(old_path) and os.path.exists(new_path):
    print("✅ Already renamed:", new_path)
else:
    os.rename(old_path, new_path)
    print("✅", new_path)
```
