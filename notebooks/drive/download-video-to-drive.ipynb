```python
#@title Download
url = "" #@param {type:"string"}
quality = "best" #@param ["best", "1080p", "720p", "480p", "audio only"]
folder = "Videos" #@param {type:"string"}

import os, subprocess
if not url:
    raise SystemExit("Enter a video URL.")
dest_dir = f"/content/drive/MyDrive/{folder}"
os.makedirs(dest_dir, exist_ok=True)
!pip install -q -U yt-dlp
if quality == "audio only":
    fmt = ["-f", "ba/b", "-x", "--audio-format", "mp3"]
elif quality == "best":
    fmt = ["-f", "bv*+ba/b", "--merge-output-format", "mp4"]
else:
    h = quality.rstrip("p")
    fmt = ["-f", f"bv*[height<={h}]+ba/b[height<={h}]", "--merge-output-format", "mp4"]
subprocess.run(["yt-dlp", *fmt, "-c", "--no-playlist",
                "-o", f"{dest_dir}/%(title).150B [%(id)s].%(ext)s", url], check=True)
```
